## 14. Missing Values

````python
# Create a small DataFrame containing missing values
missing_df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha'],
    'Age': [25, np.nan, 28, 35],
    'Salary': [50000, 70000, np.nan, 90000]
})

# Check missing values
display(missing_df.isna())

# Count missing values by column
print(missing_df.isna().sum())

# Percentage of missing values
print(missing_df.isna().mean() * 100)

# Drop rows containing missing values
display(missing_df.dropna())

# Fill all missing values with zero
display(missing_df.fillna(0))
````

````python
# Fill Age missing values with the median
missing_df['Age'] = missing_df['Age'].fillna(
    missing_df['Age'].median()
)

# Fill Salary missing values with the mean
missing_df['Salary'] = missing_df['Salary'].fillna(
    missing_df['Salary'].mean()
)

display(missing_df)

# Forward fill and backward fill
print("Forward fill:")
display(missing_df.ffill())

print("Backward fill:")
display(missing_df.bfill())
````

## 28. Merge — SQL JOIN Equivalent

````python
customers = pd.DataFrame({
    'CustomerID': [1, 2, 3],
    'Name': ['Ali', 'Sara', 'Ahmed']
})

orders = pd.DataFrame({
    'CustomerID': [1, 1, 2, 4],
    'Amount': [1000, 2000, 1500, 3000]
})

# Inner join
inner_join = pd.merge(
    customers,
    orders,
    on='CustomerID',
    how='inner'
)

display(inner_join)

# Left join
left_join = pd.merge(
    customers,
    orders,
    on='CustomerID',
    how='left'
)

display(left_join)

# Right join
right_join = pd.merge(
    customers,
    orders,
    on='CustomerID',
    how='right'
)

display(right_join)

# Outer join
outer_join = pd.merge(
    customers,
    orders,
    on='CustomerID',
    how='outer'
)

display(outer_join)
````

## 35. Train/Test Split for Machine Learning

## 36. Outlier Detection using IQR

````python
outlier_df = pd.DataFrame({
    'Salary': [40000, 45000, 50000, 52000, 55000, 60000,
               65000, 70000, 75000, 200000]
})

# Calculate first and third quartiles
Q1 = outlier_df['Salary'].quantile(0.25)
Q3 = outlier_df['Salary'].quantile(0.75)

# Interquartile range
IQR = Q3 - Q1

# Define lower and upper boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = outlier_df[
    (outlier_df['Salary'] < lower_bound) |
    (outlier_df['Salary'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

display(outliers)
````

## 40. Cumulative Operations

````python
time_df = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [100, 150, 120, 180, 200]
})

# Cumulative sum
time_df['CumulativeSales'] = time_df['Sales'].cumsum()

# Cumulative maximum
time_df['CumulativeMax'] = time_df['Sales'].cummax()

# Cumulative minimum
time_df['CumulativeMin'] = time_df['Sales'].cummin()

display(time_df)
````
